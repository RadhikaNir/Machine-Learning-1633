## K Means Clustering

```python
path = r"https://raw.githubusercontent.com/RadhikaNir/Datasets/refs/heads/main/Mall_Customers.csv"
import pandas as pd
df = pd.read_csv(path)
df.head()
```

## Perform Basic Data Quality Checks

```python
df.info()
```

```python
# Missing data
df.isna().sum()
```

```python
# Duplicates
df.duplicated().sum()
```

```python
df.columns
```

## Consider features required for X(Cluster formation)
X: age, income and spending

```python
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer

X = df[['Age','Annual Income (k$)', 'Spending Score (1-100)']]
X.head()
```

```python
X.columns = ['Age','Income','Spending']
X.head()
```

```python
num_pipe = make_pipeline(
    SimpleImputer(strategy='median'),
    StandardScaler()
).set_output(transform='pandas')

num_pipe
```

```python
X_pre = num_pipe.fit_transform(X)
X_pre.head()
```

## K Means Cluster

```python
from sklearn.cluster import KMeans
model = KMeans(n_clusters=4)
model.fit(X_pre)
```

```python
clusters4 = model.predict(X_pre)
clusters4
```

## Evaluation Metrci: Inertia

```python
model.inertia_
```

## Choosing the best cluster with the help of Elbow Method

```python
pip install yellowbrick
```

## Restart the kernel and run all the cells

```python
from yellowbrick.cluster import kelbow_visualizer
base_model = KMeans()
visualizer = kelbow_visualizer(base_model,X_pre,k=(2,15),timings=False)
visualizer.show()
```

## The value of k from the elbow method is 6.

### Rebuild the model with number of clusters as 6

```python
final_model = KMeans(n_clusters=6)
final_model.fit(X_pre)
```

```python
labels = final_model.predict(X_pre)
labels
```

## Add this labels list to the dataframe and share it with the client

```python
df['Cluster Number'] = labels
df.head()
```
