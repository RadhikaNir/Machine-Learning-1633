## 4. Data Preprocessing 
Prepare the Dataset by: 

    Separating features (X) and (Y)

```python
X = df_x
print("X records")
display(X.head())
Y = y
print("Y records")
display(Y.head())
```

    Splitting the data into 80% training and 20% testing sets.

```python
from sklearn.model_selection import train_test_split

xtrain, xtest, ytrain, ytest = train_test_split(X, Y, train_size= 0.8, random_state= 21)
print(xtrain.shape, xtest.shape)
print(ytrain.shape, ytest.shape)
```

```python
xtrain.columns
```

```python
xtest.dtypes
```

```python
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

num_pipe = make_pipeline(
    SimpleImputer(strategy='median'),
    StandardScaler()
    ).set_output(transform='pandas')
```

```python
# Fit it on training data
num_pipe.fit(xtrain)
```

```python
# transform it on testing as well as training data
xtrain_pre = num_pipe.transform(xtrain)
xtest_pre = num_pipe.transform(xtest)
```

```python
xtrain_pre.head()
```

```python
xtest_pre.head()
```

## 5. Model Training 
    Build a Logistic Regression model using the training dataset and display the model coefficients and intercept.

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression()
model.fit(xtrain_pre, ytrain)
```

```python
# model coefficients
print("Coefficient", model.coef_)
```

```python
# Intercept
print("Intercept", model.intercept_)
```

## 6. Prediction
    Predict the target classes for the testing dataset and display a comparison of the Actual and Predicted values.

```python
ypred_test = model.predict(xtest_pre)

ypred_test[:5]
ytest.head()
```
